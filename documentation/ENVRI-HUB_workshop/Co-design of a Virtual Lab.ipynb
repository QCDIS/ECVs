### Create draft components
To create a draft component for your workflow, do the following:
1) Select a notebook code cell (e.g. the `# Data storer` below) and open the NaaVRE containerizer:<br/>
<img src="./../images/containerizer.png" alt="containerizer.png" width="1000"/>
2) Press `Analyze cell`
3) Check the box `Draft`
4) Press `Create`:<br/>
<img src="./../images/create.png" alt="create.png" width="300"/>

### Compose a workflow
To use the draft components to design a workflow using the existing ECV workflow components, do the following:
1) Go to the folder `workflow`:<br/>
2) Open `workflow.naavrewf`:<br/>
3) Find the workflow component you have created in the catalog.
4) Drag and drop the component onto the canvas.<br/>
<img src="./../images/component_to_canvas.png" alt="component_to_canvas.png" width="600"/>
<br/><br/>
Create another draft component from another jupyter notebook cell before continuing with the following steps. Make sure the new draft component provides input to the existing component or uses output from the previously created component.
<br/><br/>
6) Drag and drop the second component onto the canvas.
7) Connect the ouput and input of the blocks.<br/>
<img src="./../images/connect_components.png" alt="connect_components.png" width="600"/>
